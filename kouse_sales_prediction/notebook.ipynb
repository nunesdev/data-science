##  Detecção de outliers univariados 

### a) Calcule o intervalo interquartil (IQR) para cada variável numérica e identifique os registros que se encontram abaixo de Q1 − 1.5×IQR ou acima de Q3 + 1.5×IQR.

```
# Lista de variáveis numéricas que fazem sentido para análise
numeric_cols = [
    "price", "bedrooms", "bathrooms", "sqft_living", "sqft_lot",
    "floors", "waterfront", "view", "condition", "grade",
    "sqft_above", "sqft_basement", "yr_built", "yr_renovated",
    "lat", "long", "sqft_living15", "sqft_lot15"
]

# Cálculo de Q1, Q3 e IQR para cada variável numérica
Q1 = df[numeric_cols].quantile(0.25)
Q3 = df[numeric_cols].quantile(0.75)
IQR = Q3 - Q1

# Limites inferior e superior
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Máscara de outliers: True onde o valor está fora dos limites
outlier_mask = (df[numeric_cols] < lower_bound) | (df[numeric_cols] > upper_bound)

# Registros que têm pelo menos um outlier em alguma variável numérica
df_outliers = df[outlier_mask.any(axis=1)]

df_outliers.head()
```

### b) Calcule o desvio absoluto mediano (MAD) para as mesmas variáveis e determine quais registros ultrapassam o limite de 3 unidades de MAD.

```
# Mediana de cada variável numérica
median = df[numeric_cols].median()

# MAD = mediana do desvio absoluto em relação à mediana
mad = (df[numeric_cols] - median).abs().median()

# "z-score robusto" baseado no MAD
robust_z = (df[numeric_cols] - median).abs() / mad

# Máscara: valores que ultrapassam 3 unidades de MAD
mad_outlier_mask = robust_z > 3

# Registros com pelo menos um outlier pelo critério de MAD
df_mad_outliers = df[mad_outlier_mask.any(axis=1)]

df_mad_outliers.head()
```

### c) Calcule o z-score padronizado e identifique os registros com valores absolutos maiores que 3.

```
# Média e desvio padrão de cada variável numérica
mean = df[numeric_cols].mean()
std = df[numeric_cols].std()

# z-score padronizado
z_scores = (df[numeric_cols] - mean) / std

# Máscara de outliers: |z| > 3
z_outlier_mask = z_scores.abs() > 3

# Registros com pelo menos um outlier pelo critério de z-score
df_z_outliers = df[z_outlier_mask.any(axis=1)]

df_z_outliers.head()
```

### d) Construa uma tabela comparativa com o número de outliers encontrados por método (IQR, MAD e z-score) em cada variável numérica.

```
# --- IQR ---
Q1 = df[numeric_cols].quantile(0.25)
Q3 = df[numeric_cols].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
iqr_mask = (df[numeric_cols] < lower_bound) | (df[numeric_cols] > upper_bound)
iqr_counts = iqr_mask.sum()

# --- MAD ---
median = df[numeric_cols].median()
mad = (df[numeric_cols] - median).abs().median()
mad_replaced = mad.replace(0, np.nan)
robust_z = (df[numeric_cols] - median).abs().divide(mad_replaced)
mad_mask = robust_z > 3
mad_counts = mad_mask.sum()

# --- Z-score ---
mean = df[numeric_cols].mean()
std = df[numeric_cols].std()
z_scores = (df[numeric_cols] - mean) / std
z_mask = z_scores.abs() > 3
z_counts = z_mask.sum()

# Tabela comparativa
outlier_table = pd.DataFrame({
    "IQR_outliers": iqr_counts,
    "MAD_outliers": mad_counts,
    "Zscore_outliers": z_counts
})

outlier_table
```

As variáveis com mais outliers são principalmente as ligadas a tamanho do terreno, área da casa, preço e características de qualidade/vista.

##  Inconsistências lógicas entre variáveis

### a) Defina e aplique filtros condicionais para detectar registros logicamente inconsistentes com a natureza da base de dados.

```
# Preço não pode ser menor ou igual a zero
rule_price = df["price"] <= 0

# Casas com 0 ou menos quartos 
rule_bedrooms = df["bedrooms"] <= 0

# Casas com 0 ou menos banheiros
rule_bathrooms = df["bathrooms"] <= 0

# Área construída ou área do lote menor ou igual a zero
rule_sqft_nonpositive = (df["sqft_living"] <= 0) | (df["sqft_lot"] <= 0)

# Número de andares menor ou igual a zero
rule_floors = df["floors"] <= 0

# Ano de reforma menor que o ano de construção (e diferente de 0)
rule_yr_renovated_inconsistent = (df["yr_renovated"] != 0) & (df["yr_renovated"] < df["yr_built"])

# alores fora do domínio esperado para variáveis categóricas numéricas
rule_waterfront_range = ~df["waterfront"].isin([0, 1])  
rule_view_range = ~df["view"].between(0, 4)                  
rule_condition_range = ~df["condition"].between(1, 5)        
rule_grade_range = ~df["grade"].between(1, 13)                

# Combinar todas as regras em uma máscara geral
inconsistent_mask = (
    rule_price |
    rule_bedrooms |
    rule_bathrooms |
    rule_sqft_nonpositive |
    rule_floors |
    rule_yr_renovated_inconsistent |
    rule_waterfront_range |
    rule_view_range |
    rule_condition_range |
    rule_grade_range
)

# Ver apenas os registros inconsistentes
df_inconsistent = df[inconsistent_mask]
df_inconsistent.head(10)
```

### b) Conte o número de inconsistências encontradas.

```
counts = {
    "price <= 0": int(rule_price.sum()),
    "bedrooms <= 0": int(rule_bedrooms.sum()),
    "bathrooms <= 0": int(rule_bathrooms.sum()),
    "sqft_living <= 0 or sqft_lot <= 0": int(rule_sqft_nonpositive.sum()),
    "floors <= 0": int(rule_floors.sum()),
    "yr_renovated < yr_built (e != 0)": int(rule_yr_renovated_inconsistent.sum()),
    "waterfront not in {0,1}": int(rule_waterfront_range.sum()),
    "view not in [0,4]": int(rule_view_range.sum()),
    "condition not in [1,5]": int(rule_condition_range.sum()),
    "grade not in [1,13]": int(rule_grade_range.sum()),
    "total_inconsistent": int(inconsistent_mask.sum())
}
insconsistence_table_counts = pd.DataFrame(counts, index=[0, 10])

insconsistence_table_counts
```

### c) Documente o raciocínio utilizado na escolha das regras de validação e a relevância de cada inconsistência no contexto do domínio.


As regras de validação foram definidas com base em **bom senso do domínio imobiliário** (o que é possível ou razoável para uma casa) e nos **domínios esperados das variáveis** (faixas de valores válidos).  
Cada inconsistência identificada representa um potencial **erro de registro, falta de padronização ou dado mal definido**, que pode comprometer:

-   estatísticas descritivas (médias, distribuições),
    
-   comparações entre imóveis,
    
-   e principalmente a **qualidade de modelos preditivos de preço**.  
    Por isso, documentar essas regras e os casos encontrados é uma parte importante do diagnóstico de qualidade de dados antes de qualquer decisão estratégica baseada nessa base.

## Avaliação de integridade e adequação

### a) Implemente verificações programáticas de integridade e adequação dos dados — como tipos incorretos, ausência de chaves únicas, domínio inválido de categorias ou intervalos fora do esperado

### - Verificação de tipos incorretos

```
expected_dtypes = {
    "id": "int64",
    "date": "object",        
    "price": "float64",
    "bedrooms": "int64",
    "bathrooms": "float64",
    "sqft_living": "int64",
    "sqft_lot": "int64",
    "floors": "float64",
    "waterfront": "int64",
    "view": "int64",
    "condition": "int64",
    "grade": "int64",
    "sqft_above": "int64",
    "sqft_basement": "int64",
    "yr_built": "int64",
    "yr_renovated": "int64",
    "zipcode": "int64",     
    "lat": "float64",
    "long": "float64",
    "sqft_living15": "int64",
    "sqft_lot15": "int64"
}

# Comparar tipos atuais com os tipos esperados
current_dtypes = df.dtypes.astype(str)

type_issues = []

for col, expected in expected_dtypes.items():
    if current_dtypes[col] != expected:
        type_issues.append({
            "coluna": col,
            "tipo_atual": current_dtypes[col],
            "tipo_esperado": expected
        })

pd.DataFrame(type_issues)

#Aqui verifico se o tipo de cada coluna bate com o que eu espero para a análise.
# Tipos errados podem indicar import incorreto (ex.: número lido como texto) 
# e podem atrapalhar cálculos, gráficos e modelos.
```

### - Verificação de chave única (integridade de identificação)

```
# Verifica se 'id' é única
id_is_unique = df["id"].is_unique

# Se não for única, pegar duplicados
df_id_duplicates = df[df.duplicated("id", keep=False)].sort_values("id")

id_is_unique, df_id_duplicates.head()
```

### - Verificação de domínio de categorias (valores fora do esperado)

```
issues_domain = {}

# waterfront: esperado 0 ou 1
invalid_waterfront = ~df["waterfront"].isin([0, 1])
issues_domain["waterfront_invalid"] = int(invalid_waterfront.sum())

# view: esperado de 0 a 4
invalid_view = ~df["view"].between(0, 4)
issues_domain["view_invalid"] = int(invalid_view.sum())

# condition: esperado de 1 a 5
invalid_condition = ~df["condition"].between(1, 5)
issues_domain["condition_invalid"] = int(invalid_condition.sum())

# grade: esperado de 1 a 13
invalid_grade = ~df["grade"].between(1, 13)
issues_domain["grade_invalid"] = int(invalid_grade.sum())

issues_domain
```

```
summary_integrity = {
    "tipos_incorretos_qtd": len(type_issues),
    "id_unica": id_is_unique,
    "registros_id_duplicados": len(df_id_duplicates),
    "dominio_categorias": issues_domain
}

summary_integrity
```
